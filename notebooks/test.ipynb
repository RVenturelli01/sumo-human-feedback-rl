```python
import pickle

SAVE_PATH = "/home/ricca/projects/sumo-human-feedback-rl/data_for_training/debug_dataset_full_ep.pkl"

with open(SAVE_PATH, "rb") as f:
     data = pickle.load(f)
```

```python
all_trajs = [traj for trajs in data.values() for traj in trajs]
```

```python
all_tran = [t for traj in all_trajs for t in traj]
```

```python
import numpy as np
print([np.argmax(traj.next_status) for traj in all_tran])
```

```python
STATUS_ARRIVED  = 0
STATUS_COLLIDED = 1
STATUS_OFFROAD  = 2
STATUS_TIMEOUT  = 3
STATUS_RUNNING  = 4

REWARD_MAP = {
    STATUS_ARRIVED:  20.0,
    STATUS_COLLIDED: -50.0,
    STATUS_OFFROAD:  -50.0,
    STATUS_TIMEOUT:  0.0,
    # STATUS_RUNNING: keep original value
}

from dataclasses import replace

def update_reward(transition):
    status_idx = transition.next_status.argmax().item()
    if status_idx == STATUS_RUNNING:
        return transition  # keep original reward
    new_reward = REWARD_MAP[status_idx]
    return replace(transition, true_reward=new_reward)

updated_transitions = [update_reward(t) for t in all_tran]
```

```python
rew = [t.true_reward for t in updated_transitions]
# keep only rewards between 0.5 and 20 (exclusive)
m = [r for r in rew if r < 20 and r > 0.5]
print(m)
```

```python
import pickle

SAVE_PATH = "debug_dataset.pkl"

with open(SAVE_PATH, "wb") as f:
    pickle.dump(updated_transitions, f)

print(f"Salvato in: {SAVE_PATH}")
```
